# Build a basic AI system with LangChain 

This project demonstrates how to build a simple question-answering application using **LangChain** and an **OpenAI chat model**.

The application follows the most basic LLM workflow:

```python
pip install langchain python-dotenv colorama
```

```python
from getpass import getpass
from dotenv import load_dotenv
from colorama import Fore, Style

import os

load_dotenv(override=True)  # Load environment variables from .env file

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY", "")  # Set OPENAI_API_KEY from environment variable
if not os.environ["OPENAI_API_KEY"]:
    os.environ["OPENAI_API_KEY"] = getpass("Enter OpenAI API key:")
else:
    print(f"{Fore.GREEN}OpenAI API key loaded from environment variable.{Style.RESET_ALL}")
```

**Initialize the model**

```python
from langchain.chat_models import init_chat_model


# -----------------------------
# LLM Init and Setup
# -----------------------------

# change for a more cost-effective model like gpt-4o-mini if you want to save money, but it will be slower and less capable
# model = init_chat_model("gpt-5.5")
model = init_chat_model("gpt-4o-mini")
```

**Run the Chain**

```python
response = model.invoke("What is the latitude and longitute of 19 Canterbury house, Bow Road, London E3 3AU United Kingdom?")
print(f"{Fore.CYAN}Response from model: {Style.RESET_ALL}{response}")
```
